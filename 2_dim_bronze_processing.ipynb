## BRANDS

```python
catalog = dbutils.widgets.get('catalog')
brands_source = dbutils.widgets.get('brands_source')

print(catalog, brands_source)
```

```python
# Define schema for the brands data

brands_schema = StructType([
    StructField("brand_code", StringType(), False),
    StructField("brand_name", StringType(), True),
    StructField("category_code", StringType(), True),
    
])
```

```python
# Show the path to the source

base_path = f"s3://shopvista-sv/{brands_source}/*.csv"

print(base_path)
```

```python
#Read data into dataframe and add metadata

df = (
    spark.read.format("csv")
    .option("header", True)
    .option("inferschema", True)
    .load(base_path)
    .withColumn("read_timestamp", F.current_timestamp())
    .select("*", "_metadata.file_name", "_metadata.file_path")
)

```

```python
# write data to bronze table
df.write\
    .format("delta")\
    .option("delta.enableChangeDataFeed", True)\
    .mode("overwrite")\
    .saveAsTable(f"{catalog}.{bronze_schema}.{brands_source}")
```

## DB_DATE_TABLE

```python
catalog = dbutils.widgets.get("catalog")
db_date_source = dbutils.widgets.get("db_date_source")

print(catalog, db_date_source)
```

```python
# start date and end date
start_date = '2024-01-01'
end_date = '2025-08-31'
```

```python
# Generate one row per month between start_date and end_date

df = (
    spark.sql(f"""
        SELECT explode(
            sequence(
                to_date('{start_date}'),
                to_date('{end_date}'),
                interval 1 month
            )
        ) As month_start_date
    """)
)

# Add useful analytics columns
df = (
    df 
    # Surrogate key at month grain
    .withColumn("date_key", F.date_format("month_start_date", "yyyyMM").cast("int"))
    .withColumn("year", F.year("month_start_date"))
    .withColumn("month_name", F.date_format("month_start_date", "MMMM"))
    .withColumn("month_short_name", F.date_format("month_start_date", "MMM"))
    .withColumn("quater", F.concat(F.lit("Q"), F.quarter("month_start_date")))
    .withColumn("year_quater", F.concat(F.col("year"), F.lit("-Q"), F.quarter("month_start_date")))
)
```

```python
#display(df)
```

```python
# Save as a table
df.write \
    .mode("overwrite")\
    .saveAsTable(f'{catalog}.{bronze_schema}.{db_date_source}')
```
